```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import numpy as np 

from fast_o3_iv import metrics
from fast_o3_iv.config import PROCESSED_DATA_DIR
```

```python
MODEL_KEYS = [
    "MLR-global",
    "XGB-global",
    "GPR-local",
    "MLR-local",
    "XGB-local",
]

def load_validation_results(model_key):
    """Load and concatenate all validation-set prediction results for a given model_key.
    """
    validation_dir = PROCESSED_DATA_DIR / "model_outputs" / "validation"
    matching_files = sorted(validation_dir.glob(f"*.{model_key}.parquet"))

    dfs = []
    for file_path in matching_files:
        df = pd.read_parquet(file_path)
        df["model_key"] = model_key
        dfs.append(df)

    return pd.concat(dfs, ignore_index=True)
```

```python
METRICS = ["rmse", "mae", "bias"]

results = {}

for model_key in MODEL_KEYS:
    df = load_validation_results(model_key)
    results[model_key] = {}

    for metric in METRICS:
        value = metrics.calculate_global_wmean_of_local_metric(
            df, metric, only_populated_grid_cells=True
        )
        results[model_key][metric] = np.round(value, 2)

# rows = metrics, columns = model_keys
table = pd.DataFrame(results).reindex(METRICS)

print(table)

table.to_csv(PROCESSED_DATA_DIR / "figure_data" / "Table1_validation_metrics.csv")
```
