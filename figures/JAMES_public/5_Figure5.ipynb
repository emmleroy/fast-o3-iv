```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import regionmask

from fast_o3_iv import utils, metrics
from fast_o3_iv.config import PROCESSED_DATA_DIR

from matplotlib.gridspec import GridSpec
from matplotlib.lines import Line2D

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
```

```python
MODEL_KEYS = [
    "GPR-local",
]

def load_inference_results(model_key):
    """Load and concatenate all inference-set prediction results for a given model_key.
    """

    INFERENCE_DIR = PROCESSED_DATA_DIR / "model_outputs" / "prediction"
    matching_files = sorted(INFERENCE_DIR.glob(f"*.{model_key}.parquet"))

    dfs = []
    for file_path in matching_files:
        parts = file_path.stem.split(".")
        ensemble_member = ".".join(parts[2:4])  # e.g. "wa6_ic3.002"

        df = pd.read_parquet(file_path)
        df["model_key"] = model_key
        df["ensemble_member"] = ensemble_member
        dfs.append(df)

    return pd.concat(dfs, ignore_index=True)
```

```python
for model_key in MODEL_KEYS:
    df = load_inference_results(model_key)
    ds = utils._df_to_ds(df, extra_index_cols=["ensemble_member"])
    
```
