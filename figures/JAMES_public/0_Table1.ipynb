```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import numpy as np 

from fast_o3_iv import utils, metrics
from fast_o3_iv.config import PROCESSED_DATA_DIR
```

```python
MODEL_KEYS = [
    "MLR-global",
    "XGB-global",
    "GPR-local",
    "MLR-local",
    "XGB-local",
]

METRICS = ["rmse", "mae", "bias"]

results = {}

for model_key in MODEL_KEYS:
    df = utils.load_results(model_key, "validation")
    results[model_key] = {}

    for metric in METRICS:
        value = metrics.calculate_global_wmean_of_local_metric(
            df, metric, only_populated_grid_cells=True
        )
        results[model_key][metric] = np.round(value, 2)

# rows = metrics, columns = model_keys
table = pd.DataFrame(results).reindex(METRICS)

print(table)

table.to_csv(PROCESSED_DATA_DIR / "figure_data" / "Table1_validation_metrics.csv")
```
